```python
import pandas as pd
df = pd.read_csv('/content/salse_data.csv')
df
```

```python
import numpy as np

# Pehle Revenue column ko calculate karna zaroori hai
df['Revenue'] = df['Quantity'] * df['Price']

# Revenue ko categorize karna (More than 50,000 aur Less than or equal to 50,000)
df['Revenue_Category'] = np.where(df['Revenue'] > 50000, 'More than 50,000', 'Less than or equal to 50,000')

# Nayi category ke counts dekhne ke liye
display(df[['Product', 'Revenue', 'Revenue_Category']])
print("\nCategory-wise Order Counts:")
print(df['Revenue_Category'].value_counts())
```

```python
# NumPy ka use karke maximum revenue identify karna
import numpy as np

max_revenue_val = np.max(df['Revenue'].to_numpy())
print(f"NumPy ke mutabik Maximum Revenue: ₹{max_revenue_val:,.2f}")
```

```python
# Original Revenue par 10% increment calculate karna
df['Increased_Revenue'] = df['Revenue'] * 1.10

# Original Revenue aur Nayi Increased Revenue ko side-by-side dekhna
display(df[['Product', 'Quantity', 'Price', 'Revenue', 'Increased_Revenue']].head())

# Total Revenue comparison
print(f"Original Total Revenue: ₹{df['Revenue'].sum():,.2f}")
print(f"Total Revenue after 10% Increase: ₹{df['Increased_Revenue'].sum():,.2f}")
```

```python
import pandas as pd
import numpy as np

# Revenue values ke description ke basis par hum range define karte hain
# Min Revenue is ~4,400 aur Max Revenue is 104,000.
# Hum equal-width bins ya custom ranges use kar sakte hain.
# Custom thresholds: Low (< 15,000), Medium (15,000 to 50,000), High (> 50,000)

conditions = [
    (df['Revenue'] < 15000),
    (df['Revenue'] >= 15000) & (df['Revenue'] <= 50000),
    (df['Revenue'] > 50000)
]

categories = ['Low', 'Medium', 'High']

df['Revenue_Segment'] = np.select(conditions, categories, default='Unknown')

# Displaying the resulting DataFrame
display(df[['Product', 'Revenue', 'Revenue_Segment']].sort_values(by='Revenue', ascending=False))

print("\nSegment-wise Revenue Distribution:")
print(df['Revenue_Segment'].value_counts())
```
